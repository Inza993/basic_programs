```
fun encodeMsg(msg: String, encode: (String) -> String): String {
    return encode(msg)
}

val encodeArg: (String) -> String = { msg: String -> "**$msg**" }
println(encodeMsg(msg= "secert", encode = encodeArg))
encodeMsg(msg= "secert", encode = { msg: String -> "**--$msg--**" })
```

```
val enc1: (String) -> String = { input -> input.toUpperCase() }
encodeMsg("abc", enc1)
```

```
fun enc2(input:String): String {
    return input.reversed()
}

encodeMsg("abc", ::enc2)
```

```
abstract class Food {
    abstract val kcal : Int
    abstract val name : String
    fun consume() = println("I'm eating ${name}")
}
class Pizza() : Food() {
    override val kcal = 600
    override val name = "Pizza"
}
fun main() {
    Pizza().consume()    // "I'm eating Pizza"
}
```

```
infix fun String.withPrefix(prefix: String): String {
    return "$prefix$this"
}

"Bean" withPrefix "Mr."
```

```
class PhysicsSystem {
    companion object WorldConstants {
        val gravity = 9.8
        val unit = "metric"
        fun computeForce(mass: Double, accel: Double): Double {
            return mass * accel
        }
    }
}
println(PhysicsSystem.WorldConstants.gravity)
println(PhysicsSystem.WorldConstants.computeForce(10.0, 10.0))
```
